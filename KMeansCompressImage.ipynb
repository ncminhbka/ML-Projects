```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans

#load image (load one time, after that delete this code)
#from google.colab import files
#uploaded = files.upload()

#show original image
img = mpimg.imread('girlface.jpg')
plt.imshow(img)
plt.axis('off')
plt.show()
#print(img.shape)
#reshape X to 2D arr, each pixel at each row
X = img.reshape((img.shape[0]*img.shape[1], img.shape[2]))
#print(X)
# or X = img.reshape((-1, img.shape[2]))

#clustering
kmeans = KMeans(n_clusters=10)
kmeans.fit(X)

#change every point in each kmeans.labels by its center
X = kmeans.cluster_centers_[kmeans.labels_]

#reshape back to an image
X = X.reshape((img.shape[0], img.shape[1], img.shape[2]))
#but X has float number, change all of them to int[0, 255]
X = X.astype(np.uint8)

#show compressed image
plt.imshow(X)
plt.axis('off')
plt.show()


```
